## 1.1 El problema

Disponemos de **100 mensajes** que ya han sido clasificados:

- 40 son SPAM.
- 60 son HAM.

También conocemos cuántos mensajes de cada clase contienen determinadas palabras.

| Palabra | SPAM | HAM |
|---|---:|---:|
| gratis | 18 | 2 |
| premio | 15 | 1 |
| oferta | 14 | 4 |
| reunión | 1 | 16 |
| proyecto | 2 | 18 |
| urgente | 10 | 5 |

Estos datos son ficticios y se utilizan únicamente con finalidad didáctica.

Según la unidad, Naive Bayes es un algoritmo basado en el **Teorema de Bayes** que asume que las variables de entrada son independientes entre sí. Vamos a utilizar esa idea para decidir si determinados mensajes se parecen más a los mensajes SPAM o a los HAM.

## 1.2 Variables y operaciones aritméticas

En Python podemos guardar datos en **variables**. Empezaremos almacenando el número de mensajes de cada clase.

```python
total_mensajes = 100

spam = 40
ham = 60

print("Total de mensajes:", total_mensajes)
print("Mensajes SPAM:", spam)
print("Mensajes HAM:", ham)
```

### Ejercicio guiado 1. Calcular las probabilidades iniciales

Antes de observar ninguna palabra podemos calcular qué proporción de mensajes pertenece a cada clase.

Para ello dividimos el número de mensajes de una clase entre el número total de mensajes.

```python
p_spam = spam / total_mensajes

p_ham = ham / total_mensajes

print("P(SPAM) =", p_spam)
print("P(HAM)  =", p_ham)
```

Observa algunos elementos básicos de Python:

- `=` asigna un valor a una variable.
- `/` realiza una división.
- `print()` muestra información por pantalla.
- Los comentarios comienzan por `#` y Python no los ejecuta.

En este caso obtenemos `0.4` y `0.6`, que equivalen al 40 % y 60 % respectivamente.

## 1.4 Diccionarios: organizar los datos

Necesitamos guardar las apariciones de varias palabras. Una estructura muy útil en Python es el **diccionario**.

Un diccionario relaciona una **clave** con un **valor**. En nuestro caso, la clave será una palabra y el valor será el número de mensajes en los que aparece.

```python
palabras_spam = {
    "gratis": 18,
    "premio": 15,
    "oferta": 14,
    "reunión": 1,
    "proyecto": 2,
    "urgente": 10
}

palabras_ham = {
    "gratis": 2,
    "premio": 1,
    "oferta": 4,
    "reunión": 16,
    "proyecto": 18,
    "urgente": 5
}

print(palabras_spam)
print(palabras_ham)
```

Para consultar un dato utilizamos su clave entre corchetes.

```python
print("SPAM que contienen 'gratis':", palabras_spam["gratis"])
print("HAM que contienen 'gratis':", palabras_ham["gratis"])
```

### Ejercicio guiado 3. Probabilidad de una palabra según la clase

Queremos calcular:

- P(gratis | SPAM)
- P(gratis | HAM)

Para este ejercicio dividiremos el número de mensajes de cada clase que contienen `"gratis"` entre el total de mensajes de esa clase.

```python
p_gratis_spam = palabras_spam["gratis"] / spam
p_gratis_ham = palabras_ham["gratis"] / ham

print("P(gratis | SPAM) =", p_gratis_spam)
print("P(gratis | HAM)  =", p_gratis_ham)
```

## 1.5 Funciones

Estamos repitiendo el mismo cálculo para distintas palabras. En Python podemos crear una **función** para reutilizar código.

Una función se define con `def`. Puede recibir datos mediante parámetros y devolver un resultado mediante `return`.

```python
def probabilidad_palabra(palabra, datos, total_clase):

    apariciones = datos[palabra]

    probabilidad = apariciones / total_clase

    return probabilidad

print(probabilidad_palabra("premio", palabras_spam, spam))
print(probabilidad_palabra("premio", palabras_ham, ham))
```

### Ejercicio guiado 4. Comparar una palabra

Vamos a crear una función que muestre las probabilidades de una palabra tanto para SPAM como para HAM.

```python
def comparar_palabra(palabra):
    p_en_spam = probabilidad_palabra(palabra, palabras_spam, spam)
    p_en_ham = probabilidad_palabra(palabra, palabras_ham, ham)

    print("Palabra:", palabra)
    print("P(", palabra, "| SPAM) =", p_en_spam)
    print("P(", palabra, "| HAM)  =", p_en_ham)


comparar_palabra("oferta")
```

## 1.6 Listas y bucles `for`

### Ejercicio guiado 5. Calcular el valor de un mensaje

Para clasificar `"oferta gratis"` vamos a partir de la probabilidad de cada clase y multiplicarla por las probabilidades de observar las palabras en esa clase.

En este ejemplo simplificado:

**SPAM**

`P(SPAM) × P(oferta | SPAM) × P(gratis | SPAM)`

**HAM**

`P(HAM) × P(oferta | HAM) × P(gratis | HAM)`

Esta multiplicación refleja el supuesto de independencia que la unidad atribuye a Naive Bayes.

```python
p_oferta_spam = probabilidad_palabra("oferta", palabras_spam, spam)
p_gratis_spam = probabilidad_palabra("gratis", palabras_spam, spam)

p_oferta_ham = probabilidad_palabra("oferta", palabras_ham, ham)
p_gratis_ham = probabilidad_palabra("gratis", palabras_ham, ham)

valor_spam = p_spam * p_oferta_spam * p_gratis_spam
valor_ham = p_ham * p_oferta_ham * p_gratis_ham

print("Valor para SPAM:", valor_spam)
print("Valor para HAM:", valor_ham)
```

## 1.7 Condicionales: tomar una decisión

Una vez calculados ambos valores necesitamos decidir cuál es mayor.

Python utiliza `if`, `elif` y `else` para ejecutar código dependiendo de una condición.

```python
if valor_spam > valor_ham:
    print("El mensaje se clasifica como SPAM")
elif valor_ham > valor_spam:
    print("El mensaje se clasifica como HAM")
else:
    print("Se ha producido un empate")
```

# Apartado 2: Crea tu primer clasificador

##Ejercicio 1: Diseña una función para integrar todo lo visto hasta ahora.

La función recibirá una lista de palabras, calculará un valor para SPAM y otro para HAM y devolverá la clase con el valor más alto.

```python
def clasificar_mensaje(palabras):

    ## Añade aquí tu código

    if valor_spam > valor_ham:
        return "SPAM"
    elif valor_ham > valor_spam:
        return "HAM"
    else:
        return "EMPATE"


mensaje = ["oferta", "gratis"]
resultado = clasificar_mensaje(mensaje)

print("Clasificación:", resultado)
```

Ya podemos reutilizar nuestra función. Esta es una de las principales ventajas de las funciones: escribimos el procedimiento una vez y después lo aplicamos a distintos datos.

```python
mensajes = [
    ["premio", "gratis"],
    ["reunión", "proyecto"],
    ["oferta", "urgente"],
    ["proyecto", "urgente"]
]

for mensaje in mensajes:
    print("\nMensaje:", mensaje)
    resultado = clasificar_mensaje(mensaje)
    print("Resultado:", resultado)
```
